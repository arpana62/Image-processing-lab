```python
#Perform AND, OR, and NOT operations on an image.
# Import libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Upload/read image
image_path = "/content/nature.jpg"
image = cv2.imread(image_path)

# Convert image to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Create binary image
threshold_value, binary = cv2.threshold(
    gray, 127, 255, cv2.THRESH_BINARY
)

# Create inverse mask
mask = cv2.bitwise_not(binary)

# AND operation
and_image = cv2.bitwise_and(gray, gray, mask=binary)

# OR operation
or_image = cv2.bitwise_or(gray, mask)

# NOT operation
not_image = cv2.bitwise_not(gray)

# Display results
# Display results
plt.figure(figsize=(18, 5))

plt.subplot(1, 5, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 5, 2)
plt.imshow(binary, cmap="gray")
plt.title("Binary Image")
plt.axis("off")

plt.subplot(1, 5, 3)
plt.imshow(and_image, cmap="gray")
plt.title("AND Operation")
plt.axis("off")

plt.subplot(1, 5, 4)
plt.imshow(or_image, cmap="gray")
plt.title("OR Operation")
plt.axis("off")

plt.subplot(1, 5, 5)
plt.imshow(not_image, cmap="gray")
plt.title("NOT Operation")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
#Perform the four arithmetic operations on an image.
# Import libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Upload/read image
image_path = "/content/nature.jpg"
image = cv2.imread(image_path)

# Convert image to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Create second image by flipping the original image
image2 = cv2.flip(gray, 1)

# Addition
addition = cv2.add(gray, image2)

# Subtraction
subtraction = cv2.subtract(gray, image2)

# Multiplication
multiplication = cv2.multiply(
    gray, image2, scale=1/255
)

# Division
division = cv2.divide(
    gray, image2 + 1, scale=255
)
# Display results
plt.figure(figsize=(18, 5))

plt.subplot(1, 6, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 6, 2)
plt.imshow(image2, cmap="gray")
plt.title("Second Image")
plt.axis("off")

plt.subplot(1, 6, 3)
plt.imshow(addition, cmap="gray")
plt.title("Addition")
plt.axis("off")

plt.subplot(1, 6, 4)
plt.imshow(subtraction, cmap="gray")
plt.title("Subtraction")
plt.axis("off")

plt.subplot(1, 6, 5)
plt.imshow(multiplication, cmap="gray")
plt.title("Multiplication")
plt.axis("off")

plt.subplot(1, 6, 6)
plt.imshow(division, cmap="gray")
plt.title("Division")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
#Sobel operator
# Import libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Upload/read image
image_path = "/content/nature.jpg"
image = cv2.imread(image_path)

# Convert image to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Apply Sobel X
sobel_x = cv2.Sobel(
    gray, cv2.CV_64F, 1, 0, ksize=3
)

# Apply Sobel Y
sobel_y = cv2.Sobel(
    gray, cv2.CV_64F, 0, 1, ksize=3
)

# Convert to absolute values
sobel_x = cv2.convertScaleAbs(sobel_x)
sobel_y = cv2.convertScaleAbs(sobel_y)

# Combine Sobel X and Y
sobel_combined = cv2.addWeighted(
    sobel_x, 0.5, sobel_y, 0.5, 0
)

# Display results
plt.figure(figsize=(15, 5))

plt.subplot(1, 4, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 4, 2)
plt.imshow(sobel_x, cmap="gray")
plt.title("Sobel X")
plt.axis("off")

plt.subplot(1, 4, 3)
plt.imshow(sobel_y, cmap="gray")
plt.title("Sobel Y")
plt.axis("off")

plt.subplot(1, 4, 4)
plt.imshow(sobel_combined, cmap="gray")
plt.title("Combined Sobel")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
#Weighted average filter
# Import libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Upload/read image
image_path = "/content/nature.jpg"
image = cv2.imread(image_path)

# Convert image to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Define weighted average mask
kernel = np.array([
    [1, 2, 1],
    [2, 4, 2],
    [1, 2, 1]
], dtype=np.float32) / 16

# Apply weighted average filter
weighted_filter = cv2.filter2D(
    gray, -1, kernel
)

# Display results
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(weighted_filter, cmap="gray")
plt.title("Weighted Average Filter")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
#Sharpening filter
# Import libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Upload/read image
image_path = "/content/nature.jpg"
image = cv2.imread(image_path)

# Convert image to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Define sharpening mask
kernel = np.array([
    [0, -1, 0],
    [-1, 5, -1],
    [0, -1, 0]
])

# Apply sharpening filter
sharpened = cv2.filter2D(
    gray, -1, kernel
)

# Display results
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(sharpened, cmap="gray")
plt.title("Sharpened Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
#Min, max, mean, and median filters
# Import libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Upload/read image
image_path = "/content/nature.jpg"
image = cv2.imread(image_path)

# Convert image to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Create filter mask
kernel = np.ones((3, 3), np.uint8)

# Apply Min filter
min_filter = cv2.erode(gray, kernel)

# Apply Max filter
max_filter = cv2.dilate(gray, kernel)

# Apply Mean filter
mean_filter = cv2.blur(gray, (3, 3))

# Apply Median filter
median_filter = cv2.medianBlur(gray, 3)

# Display results
plt.figure(figsize=(18, 5))

plt.subplot(1, 5, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 5, 2)
plt.imshow(min_filter, cmap="gray")
plt.title("Min Filter")
plt.axis("off")

plt.subplot(1, 5, 3)
plt.imshow(max_filter, cmap="gray")
plt.title("Max Filter")
plt.axis("off")

plt.subplot(1, 5, 4)
plt.imshow(mean_filter, cmap="gray")
plt.title("Mean Filter")
plt.axis("off")

plt.subplot(1, 5, 5)
plt.imshow(median_filter, cmap="gray")
plt.title("Median Filter")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
#the Laplacian mask
# Import libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Upload/read image
image_path = "/content/nature.jpg"
image = cv2.imread(image_path)

# Convert image to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Define Laplacian mask
kernel = np.array([
    [0, 1, 0],
    [1, -4, 1],
    [0, 1, 0]
])

# Apply Laplacian mask
laplacian = cv2.filter2D(
    gray, cv2.CV_64F, kernel
)

# Convert to absolute values
laplacian = cv2.convertScaleAbs(laplacian)

# Display results
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(laplacian, cmap="gray")
plt.title("Laplacian Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
#high-boost mask
# Import libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Upload/read image
image_path = "/content/nature.jpg"
image = cv2.imread(image_path)

# Convert image to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Apply Gaussian blur
blurred = cv2.GaussianBlur(
    gray, (5, 5), 0
)

# Set high-boost factor
A = 1.5

# Apply high-boost filtering
high_boost = cv2.addWeighted(
    gray, A, blurred, -(A - 1), 0
)

# Display results
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(blurred, cmap="gray")
plt.title("Blurred Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(high_boost, cmap="gray")
plt.title("High-Boost Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
